```python
import ipywidgets as widgets
```

```python
name = widgets.Text(description="Name:", placeholder="Enter your full name")
student_id = widgets.Text(description="Student ID:", placeholder="Enter your ID")
age = widgets.BoundedIntText(description="Age:", min=10, max=100, value=18)
gender = widgets.Dropdown(
    options=["Select", "Male", "Female", "Other"],
    value="Select",
    description="Gender:"
)
```

```python
email = widgets.Text(description="Email:", placeholder="Enter your email")
submit_button = widgets.Button(description="Submit", button_style='success')
output = widgets.Output()
```

```python
submit_button.on_click(on_submit_clicked)
```

```python
form_items = widgets.VBox([name, student_id, age, gender, email, submit_button, output])
display(form_items)
```

```python
import sqlite3
```

```python
conn = sqlite3.connect("students.db")
cursor = conn.cursor()
```

```python
cursor.execute('''
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    student_id TEXT,
    age INTEGER,
    gender TEXT,
    email TEXT
)
''')
conn.commit()
conn.commit()
conn.close()
```

```python
name = widgets.Text(description="Name:", placeholder="Enter your full name")
student_id = widgets.Text(description="Student ID:", placeholder="Enter your ID")
age = widgets.BoundedIntText(description="Age:", min=10, max=100, value=18)
gender = widgets.Dropdown(options=["Select", "Male", "Female", "Other"], value="Select", description="Gender:")
email = widgets.Text(description="Email:", placeholder="Enter your email")
submit_button = widgets.Button(description="Submit", button_style='success')
output = widgets.Output()
def on_submit_clicked(b):
    with output:
        clear_output()
```

```python
def insert_to_db(name, student_id, age, gender, email):
    conn = sqlite3.connect("students.db")
    cursor = conn.cursor()
    cursor.execute('''
        INSERT INTO students (name, student_id, age, gender, email)
        VALUES (?, ?, ?, ?, ?)
    ''', (name, student_id, age, gender, email))
    conn.commit()
    conn.close()
```

```python
def on_submit_clicked(b):
    with output:
        clear_output()
        if gender.value == "Select":
            print("⚠️ Please select a gender.")
            return
        insert_to_db(name.value, student_id.value, age.value, gender.value, email.value)
        print("✅ Student details successfully submitted to the database!")

submit_button.on_click(on_submit_clicked)
```

```python
form_items = widgets.VBox([name, student_id, age, gender, email, submit_button, output])
display(form_items)
```
