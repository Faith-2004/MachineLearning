```python
import pandas as pd
import random

# Load datasets
classes_df = pd.read_csv("classes_dataset.csv")
teachers_df = pd.read_csv("teachers_dataset.csv")
time_slots_df = pd.read_csv("time_slots_dataset.csv")

classes = classes_df["class_name"].tolist()
time_slots = time_slots_df["time_slot"].tolist()

# Build subject-teacher map
subject_teacher_map = {}
for _, row in teachers_df.iterrows():
    teacher = row["name"]
    subjects = [s.strip() for s in row["subjects"].split(",")]
    for subject in subjects:
        subject_teacher_map.setdefault(subject, []).append(teacher)

# Define weekdays
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

# Initialize teacher schedule
teacher_schedule = {
    teacher: {day: set() for day in days}
    for teacher in teachers_df["name"]
}

# Initialize timetable per class and day
timetable = {cls: {day: [] for day in days} for cls in classes}

# Clean malformed time slots (ensure all have space between day and time)
cleaned_time_slots = [ts for ts in time_slots if ' ' in ts]

# Fill timetable
for cls in classes:
    for slot in cleaned_time_slots:
        day, _, period = slot.partition(' ')
        if day not in days:
            continue

        # Handle Breaks
        if period == "9:20-9:30":
            timetable[cls][day].append("☕ Short Break")
            continue
        elif period == "10:50-11:20":
            timetable[cls][day].append("☕ Tea Break")
            continue
        elif period == "12:40-1:20":
            timetable[cls][day].append("🍱 Lunch Break")
            continue
        elif period == "5:00-6:00":
            timetable[cls][day].append("🏃 Games")
            continue

        # Assign subject randomly
        subject = random.choice(list(subject_teacher_map.keys()))
        available_teachers = subject_teacher_map[subject]

        assigned = False
        for teacher in available_teachers:
            if period not in teacher_schedule[teacher][day]:
                timetable[cls][day].append(f"{period} - {subject} ({teacher})")
                teacher_schedule[teacher][day].add(period)
                assigned = True
                break

        if not assigned:
            timetable[cls][day].append(f"{period} - 🕓 Free Period")
```

```python
for cls in timetable:
    print(f"\n📘 Timetable for {cls}")
    for day in days:
        print(f"\n🗓️ {day}")
        for entry in timetable[cls][day]:
            print(f" - {entry}")
    break  # remove this break to show all classes
```
