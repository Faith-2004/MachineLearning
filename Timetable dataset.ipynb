```python
import numpy as np 
```

```python
timetable = {cls: {} for cls in classes_df[classes]}
```

```python
import random
```

```python
teacher_schedule = {t["name"]: [] for t in teacher}  # Create a schedule for each teacher by name
for cls in classes:
    used_slots = []
    for subj in subject_df["subjects"]:  # Iterate over subjects from the subject dataset
        available_slots = [slot for slot in time_slots if slot not in used_slots]
        random.shuffle(available_slots)
        for slot in available_slots:
            # Find a teacher who can teach the subject
            for t in teacher:
                if subj in t["subject"]:
                    teacher_name = t["name"]
                    if slot not in teacher_schedule[teacher_name]:
                        timetable[cls["class_name"]][slot] = f"{subj} ({teacher_name})"
                        used_slots.append(slot)
                        teacher_schedule[teacher_name].append(slot)
                        break
            else:
                continue
            break
```

```python
for cls, schedule in timetable.items():
    print(f"\n📘 {cls} Timetable")
    for slot, lesson in schedule.items():
        print(f"{slot}: {lesson}")
```
